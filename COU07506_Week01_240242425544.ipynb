```python

import numpy as np
import matplotlib.pyplot as plt
from scipy import signal
from scipy.io import wavfile

plt.rcParams["figure.figsize"] = (9, 3.5)
```

```python

n = np.arange(-5, 11)

delta = (n == 0).astype(float)

step = (n >= 0).astype(float)

ramp = n * step

expo = 0.8**n * step
```

```python

fig, axes = plt.subplots(1, 4, figsize=(13, 2.6))

for ax, x, title in zip(
    axes,
    [delta, step, ramp, expo],
    ["δ(n)", "u(n)", "r(n)", "(0.8)^n u(n)"]
):
    ax.stem(n, x, basefmt="k-")
    ax.set_title(title)
    ax.set_xlabel("n")
    ax.set_ylabel("Amplitude")

plt.tight_layout()
plt.show()
```

```python

n = np.arange(-5, 11)

values = [0.5, -0.8, 1.1]

fig, axes = plt.subplots(1, 3, figsize=(12, 3))

for ax, a in zip(axes, values):
    x = (a**n) * (n >= 0)
    ax.stem(n, x)
    ax.set_title(f"a = {a}")
    ax.set_xlabel("n")
    ax.set_ylabel("Amplitude")

plt.tight_layout()
plt.show()
```

```python

n = np.arange(0, 41)

x1 = np.cos(2*np.pi*n/8)
x2 = np.cos(2*np.pi*n/7.5)

plt.figure(figsize=(10, 4))

plt.plot(n, x1, "o-", label="cos(2πn/8)")
plt.plot(n, x2, "s-", label="cos(2πn/7.5)")

plt.title("Comparison of Two Discrete-Time Cosine Signals")
plt.xlabel("Sample index (n)")
plt.ylabel("Amplitude")
plt.legend()
plt.grid()

plt.show()
```

```python

import librosa
import librosa.display
import numpy as np
import matplotlib.pyplot as plt

# Load the audio file
fname = "/content/light audio.aac"

x, fs = librosa.load(fname, sr=None, mono=True)

# Normalize the audio signal
x = x / np.max(np.abs(x))

# Generate time axis
t = np.arange(len(x)) / fs

# Display audio information
print(f"Sampling rate = {fs} Hz")
print(f"Duration = {len(x)/fs:.2f} seconds")
print(f"Total samples = {len(x)}")

# Plot the audio waveform
plt.figure(figsize=(9, 3.5))

plt.plot(t, x, color="blue", alpha=0.7)

plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.title("My Voice Recording")
plt.grid(True)

plt.tight_layout()
plt.show()
```
